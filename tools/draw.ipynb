```python
import numpy as np
import matplotlib.pyplot as plt

# 数据
audio_features = ['mfcc', 'msclap', 'mulan', 'mulan-token', 'muq-token']
afm_results = [81.52, 81.57, 82.65, 82.69, 82.86]
masknet_results = [82.54, 83.05, 83.45, 83.30, 83.66]
dcnv2_results = [81.89, 82.51, 82.92, 82.78, 83.10]

plt.figure(figsize=(14, 8))
plt.grid(True, alpha=0.3, axis='y')

# 设置柱状图的位置
models = ['AFM', 'MaskNet', 'DCNv2']
x = np.arange(len(models))  # 模型的位置
width = 0.15  # 柱状图的宽度

# 绘制五种音频特征的柱状图
bars1 = plt.bar(x - 2*width, [afm_results[0], masknet_results[0], dcnv2_results[0]], 
                width, label='mfcc', color='purple', alpha=0.8)
bars2 = plt.bar(x - width, [afm_results[1], masknet_results[1], dcnv2_results[1]], 
                width, label='msclap', color='skyblue', alpha=0.8)
bars3 = plt.bar(x, [afm_results[2], masknet_results[2], dcnv2_results[2]], 
                width, label='mulan', color='lightgreen', alpha=0.8)
bars4 = plt.bar(x + width, [afm_results[3], masknet_results[3], dcnv2_results[3]], 
                width, label='mulan-token', color='orange', alpha=0.8)
bars5 = plt.bar(x + 2*width, [afm_results[4], masknet_results[4], dcnv2_results[4]], 
                width, label='muq-token', color='lightcoral', alpha=0.8)

# 添加数值标签在柱状图上方
def add_value_labels(bars, values):
    for bar, value in zip(bars, values):
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2., height + 0.05,
                f'{value:.2f}', ha='center', va='bottom', fontsize=14)

add_value_labels(bars1, [afm_results[0], masknet_results[0], dcnv2_results[0]])
add_value_labels(bars2, [afm_results[1], masknet_results[1], dcnv2_results[1]])
add_value_labels(bars3, [afm_results[2], masknet_results[2], dcnv2_results[2]])
add_value_labels(bars4, [afm_results[3], masknet_results[3], dcnv2_results[3]])
add_value_labels(bars5, [afm_results[4], masknet_results[4], dcnv2_results[4]])

# 设置图形属性
plt.xlabel('Models', fontsize=22)
plt.ylabel('AUC Score', fontsize=22)
# plt.title('Model Performance Comparison Across Different Audio Features', fontsize=20)
plt.xticks(x, models, fontsize=20)
plt.yticks(fontsize=20)

# 设置y轴范围
plt.ylim(81, 84)

# 添加网格


# 添加图例
plt.legend(fontsize=20, loc='upper left')

# 调整布局
plt.tight_layout()

# 保存图片
plt.savefig('../pics/model_comparison_audio_features.eps', format='eps', dpi=600, bbox_inches='tight')
# plt.savefig('../pics/model_comparison_audio_features.png', format='png', dpi=300, bbox_inches='tight')

# 显示图形
plt.show()
```
