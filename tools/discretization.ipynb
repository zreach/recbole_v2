```python
import pandas as pd
# m4a_item = pd.read_csv('/home/zhangyk/zhouyz/rec/recbole_v2/dataset/m4a/m4a.item', sep='\t')
lfm1b_item = pd.read_csv('/home/zhangyk/zhouyz/rec/recbole_v2/dataset/lfm1b-fil/lfm1b-fil.item', sep='\t')
lfm2b_item = pd.read_csv('/home/zhangyk/zhouyz/rec/recbole_v2/dataset/lfm2b-fil/lfm2b-fil.item', sep='\t')
```

```python
import pandas as pd
import numpy as np

def discretize_float_columns(df, bucket_dict):
    """
    对DataFrame中的float类型列进行离散化
    
    参数:
    df: DataFrame，列名格式为 "name:type"
    bucket_dict: dict，格式为 {"name": num_buckets}
    
    返回:
    DataFrame: 离散化后的DataFrame
    """
    df_copy = df.copy()
    
    for col in df_copy.columns:
        # 解析列名
        if ':' in col:
            name, col_type = col.split(':', 1)
            
            # 检查是否为float类型且在bucket_dict中
            if col_type == 'float' and name in bucket_dict:
                num_buckets = bucket_dict[name]
                
                # 进行离散化
                discretized_values = pd.cut(df_copy[col], 
                                          bins=num_buckets, 
                                          labels=False, 
                                          duplicates='drop')
                
                # 创建新列名
                new_col_name = f"{name}:token"
                
                # 替换列
                df_copy[new_col_name] = discretized_values
                df_copy.drop(col, axis=1, inplace=True)
    
    return df_copy
```

```python
bucket_dict = {
    'popularity': 10,
    'release': 5,
    'danceability:': 10,
    'energy:': 10,
}

result = discretize_float_columns(m4a_item, bucket_dict)
```

```python
result.to_csv('/home/zhangyk/zhouyz/rec/recbole_v2/dataset/m4a/m4a_discretized.item', sep='\t', index=False)
```

```python
lfm1b_user = pd.read_csv('/home/zhangyk/zhouyz/rec/recbole_v2/dataset/lfm1b-fil/lfm1b-fil.user', sep='\t')
lfm1b_bucket_dict = {
    'age': 10,
    'playcount': 20,
}
result = discretize_float_columns(lfm1b_user, lfm1b_bucket_dict)
result.to_csv('/home/zhangyk/zhouyz/rec/recbole_v2/dataset/lfm1b-fil/lfm1b_discretized.user', sep='\t', index=False)
```
