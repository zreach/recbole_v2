```python
import pickle 
from sklearn.cluster import KMeans
import pandas as pd


feature_path = "/home/zhangyk/zhouyz/rec/recbole_v2/dataset/lfm1b-fil/lfm1b-muq.pkl"
with open(feature_path, 'rb') as f:
    lfm1b_muq = pickle.load(f)

# feature_path = "/home/zhangyk/zhouyz/rec/recbole_v2/dataset/lfm1b-fil/lfm1b-mulan.pkl"
# with open(feature_path, 'rb') as f:
#     lfm1b_mulan = pickle.load(f)
```

```python
lfm1b_muq_layer = {k: v[1, :] for k, v in lfm1b_muq.items()}
vectors = np.array(list(lfm1b_muq_layer.values()))
vectors.shape # (95059, 1024)
```

```python
import os
os.environ['OPENBLAS_NUM_THREADS'] = '1'
def cluster_features(features, n_clusters=10):
    keys = list(features.keys())
    vectors = np.array(list(features.values()))
    print("开始聚类...")
    kmeans = KMeans(n_clusters=n_clusters, random_state=42, n_init='auto')

    kmeans.fit(vectors)
    labels = kmeans.labels_

    clustered_results = {key: label for key, label in zip(keys, labels)}

    return clustered_results
```

```python
import os
os.environ['OPENBLAS_NUM_THREADS'] = '1'
# 设置不同的聚类数
cluster_numbers = [4, 8, 16, 32, 64]
output_dir = "/home/zhangyk/zhouyz/rec/recbole_v2/dataset/lfm1b-fil/kmeans"
os.makedirs(output_dir, exist_ok=True)
item_ids = list(lfm1b_muq.keys())
results = {}
for n_clusters in cluster_numbers:
    print(f"===== 开始处理聚类数: {n_clusters} =====")
    # 创建一个空的 DataFrame 来存储所有层的聚类结果
    all_clusters_df = pd.DataFrame(index=item_ids)

    # 聚类每一层特征
    for layer in range(12):
        print(f"聚类第 {layer} 层特征，聚类数目为 {n_clusters}")
        lfm1b_muq_layer = {k: v[layer, :] for k, v in lfm1b_muq.items()}  # 获取第 layer 层的特征

        clustered_results = cluster_features(lfm1b_muq_layer, n_clusters=n_clusters)
        # 将当前层的结果作为新的一列添加到 DataFrame
        all_clusters_df[f'layer_{layer}:token'] = pd.Series(clustered_results)

    # 为每个聚类数生成不同的文件名
    output_filename = f"lfm1b-muq-alllayers-clusters-{n_clusters}.csv"
    output_path = os.path.join(output_dir, output_filename)
    all_clusters_df.reset_index(inplace=True)
    all_clusters_df.rename(columns={'index': 'tracks_id:token'}, inplace=True)
    all_clusters_df.to_csv(output_path)
    
    results[n_clusters] = all_clusters_df

    print(f"聚类数为 {n_clusters} 的结果已保存到 {output_path}")
    print(f"===== 完成处理聚类数: {n_clusters} =====\n")

print("所有聚类任务完成。")
```
