```python
import pandas as pd
import numpy as np
import pickle
import os
```

```python
with open("../dataset/msd-msclap.pkl", "rb") as f:
    msd_msclap = pickle.load(f)

with open("../dataset/msd-mulan.pkl", "rb") as f:
    msd_mulan = pickle.load(f)

with open("../dataset/msd-mfcc.pkl", "rb") as f:
    msd_mfcc = pickle.load(f)
```

```python
item_df = pd.read_csv("../dataset/lfm2b-fil/lfm2b-fil.item", delimiter="\t")
```

```python
id2msd = pd.read_csv('../dataset/lfm2b-fil/id_msd.csv', sep='\t')
id2msd_dict = dict(zip(id2msd['track_id'], id2msd['key']))
msd2id_dict = dict(zip(id2msd['key'], id2msd['track_id']))
```

```python
from tqdm import tqdm
lfm_msclap_dict = {}
for k, v in tqdm(msd_msclap.items()):
    if k in msd2id_dict.keys():
        lfm_msclap_dict[str(msd2id_dict[k])] = v

lfm_mulan_dict = {}
for k, v in tqdm(msd_mulan.items()):
    if k in msd2id_dict.keys():
        lfm_mulan_dict[str(msd2id_dict[k])] = v

lfm_mfcc_dict = {}
for k, v in tqdm(msd_mfcc.items()):
    if k in msd2id_dict.keys():
        lfm_mfcc_dict[str(msd2id_dict[k])] = v

print(len(lfm_msclap_dict))
print(len(lfm_mulan_dict))
print(len(lfm_mfcc_dict))
print(lfm_msclap_dict['19145631'].shape)
print(lfm_mulan_dict['19145631'].shape)
print(lfm_mfcc_dict['19145631'].shape)
```

```python
with open("../dataset/lfm2b-fil/lfm2b-msclap.pkl", "wb") as f:
    pickle.dump(lfm_msclap_dict, f)
with open("../dataset/lfm2b-fil/lfm2b-mulan.pkl", "wb") as f:
    pickle.dump(lfm_mulan_dict, f)
with open("../dataset/lfm2b-fil/lfm2b-mfcc.pkl", "wb") as f:
    pickle.dump(lfm_mfcc_dict, f)
```
